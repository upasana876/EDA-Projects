**5.5 Preparing the Time Period :-**

## **6. Exploratory Data Analysis (EDA)**

Exploratory Data Analysis helps us understand revenue patterns, quarterly variations, and changes in sales over time. We will use summary statistics and visualizations to identify important trends in Adidas quarterly revenue.

**6.1. Descriptive Statistical Analysis :-**

Descriptive statistical analysis helps summarize the main characteristics of quarterly revenue data. It provides information about the average, minimum, maximum, and variation in revenue, helping us understand the overall sales performance.

```python
df['Revenue'].describe()
```

```python
print("Average Revenue:", df['Revenue'].mean())
print("Median Revenue:", df['Revenue'].median())
print("Minimum Revenue:", df['Revenue'].min())
print("Maximum Revenue:", df['Revenue'].max())
```

**Key Observation:** The dataset contains 88 quarterly revenue records. The average revenue is 3,302.03, while the median is 3,162. Revenue ranges from 1,078 to 6,410, indicating considerable variation across the observed quarters.

**6.2. Quarterly Revenue Trend :-**

A line chart is used to visualize revenue changes across quarters. It helps identify overall trends, fluctuations, and periods of increasing or decreasing revenue over time.

```python
plt.figure(figsize=(14, 6))

plt.plot(df['Time Period'], df['Revenue'], marker='o')

plt.title('Adidas Quarterly Revenue Trend')
plt.xlabel('Time Period')
plt.ylabel('Revenue')
plt.xticks(rotation=90)
plt.grid(True)
plt.tight_layout()
plt.show()
```

**Key Observation:** The quarterly revenue trend shows an overall increase from 2000 to 2019, despite periodic fluctuations. Revenue reaches its highest level around 2019, followed by a sharp decline around 2020 and a subsequent recovery. This suggests long-term growth with noticeable short-term variations.

**6.3. Year-wise Revenue Analysis :-**

Year-wise revenue analysis helps compare total revenue across different years. Grouping quarterly revenue by year allows us to identify high-performing and low-performing years.

```python
yearly_revenue = df.groupby('Year')['Revenue'].sum()

yearly_revenue
```

```python
plt.figure(figsize=(14, 6))

yearly_revenue.plot(kind='bar')

plt.title('Year-wise Total Revenue')
plt.xlabel('Year')
plt.ylabel('Total Revenue')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

**Key Observation:** Year-wise analysis shows a substantial increase in total revenue from 5,835 in 2000 to a peak of 23,640 in 2019. Revenue declined in 2020 but recovered in 2021 to 21,234. Overall, the dataset indicates long-term revenue growth with a temporary decline in 2020.

**6.4. Quarter-wise Revenue Analysis :-**

Quarter-wise analysis compares average revenue across Q1, Q2, Q3, and Q4. It helps identify quarterly patterns and determine whether revenue tends to be higher or lower during particular quarters.

***Average revenue per quarter :***

```python
quarterly_revenue = df.groupby('Quarter')['Revenue'].mean()

quarterly_revenue = quarterly_revenue.reindex(['Q1', 'Q2', 'Q3', 'Q4'])

quarterly_revenue
```

***Bar chart :***

```python
plt.figure(figsize=(8, 5))

quarterly_revenue.plot(kind='bar')

plt.title('Average Revenue by Quarter')
plt.xlabel('Quarter')
plt.ylabel('Average Revenue')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

**Key Observation:** Quarter-wise analysis shows that Q3 has the highest average revenue (3,703.77), while Q2 has the lowest (3,039.18). This indicates that revenue varies across quarters, with Q3 performing comparatively better in this dataset.

**6.5. Year-wise Revenue Trend :-**

A line chart of annual revenue helps visualize long-term growth and decline. It makes it easier to identify major changes in yearly performance.

```python
plt.figure(figsize=(12, 5))

plt.plot(yearly_revenue.index, yearly_revenue.values, marker='o')

plt.title('Annual Revenue Trend')
plt.xlabel('Year')
plt.ylabel('Total Revenue')
plt.xticks(yearly_revenue.index, rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

**Key Observation:** The annual revenue trend shows substantial long-term growth, particularly from 2006 onward. Revenue reached its highest level in 2019, declined in 2020, and recovered in 2021. Despite short-term fluctuations, the overall trend across the dataset is upward.

**6.6. Revenue Distribution Analysis :-**

Revenue distribution analysis helps us understand how frequently different revenue values occur and whether the data is concentrated around a particular range. A histogram provides a visual summary of the distribution.

***Histogram :***

```python
plt.figure(figsize=(9, 5))

plt.hist(df['Revenue'], bins=10, edgecolor='black')

plt.title('Distribution of Quarterly Revenue')
plt.xlabel('Revenue')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

***Box Plot :***

```python
plt.figure(figsize=(9, 4))

sns.boxplot(x=df['Revenue'])

plt.title('Box Plot of Quarterly Revenue')
plt.xlabel('Revenue')
plt.tight_layout()
plt.show()
```

**Key Observation:** The revenue histogram shows considerable variation across quarterly observations. The box plot indicates a median revenue of 3,162, with the middle 50% of values approximately between 1,864 and 4,458. No clearly visible outlier points appear in the box plot, although this can be verified using the IQR method.

**6.7. Outlier Detection Using the IQR Method :-**

The Interquartile Range (IQR) method identifies potential outliers by calculating the range between the first and third quartiles. Values below the lower bound or above the upper bound are flagged as potential outliers.

```python
Q1 = df['Revenue'].quantile(0.25)
Q3 = df['Revenue'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['Revenue'] < lower_bound) |
    (df['Revenue'] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))

outliers
```

**Key Observation:** The IQR method identified zero potential outliers in the quarterly revenue data. All revenue values fall within the calculated lower and upper bounds. Therefore, no observations need to be removed based on this outlier detection method.
