## **5. Data Cleaning**

Data cleaning helps identify missing values, duplicate records, and invalid or unusual values. In this dataset, zero values in certain health measurements are also examined before deciding how to handle them.

**5.1 Checking Missing Values :-**

The `isnull().sum()` function identifies the number of missing values in each column.

```python
df.isnull().sum()
```

**5.2 Checking Duplicate Records :-**

The `duplicated().sum()` function counts the exact duplicate rows in the dataset.

```python
df.duplicated().sum()
```

**5.3 Checking Zero Values :-**

Zero values are examined in health-related columns to identify potentially invalid or missing measurements. Zero values in `Pregnancies` and `Outcome` can be valid, so they are treated differently.

```python
cols = ['Glucose', 'BloodPressure', 'SkinThickness',
        'Insulin', 'BMI']

df[cols].eq(0).sum()
```

**5.4 Checking Data Types :-**

The data types of all columns are checked to ensure that the variables are stored in appropriate formats for analysis.

```python
df.dtypes
```

**5.5 Final Data Quality Check :-**

The dataset is examined to summarize missing values, duplicate records, and potentially invalid zero measurements before applying any cleaning operations.

```python
print("Dataset Shape:", df.shape)
print("\nMissing Values:\n", df.isnull().sum())
print("\nDuplicate Rows:", df.duplicated().sum())
print("\nZero Values in Health Measurements:\n",
      df[cols].eq(0).sum())
```

**5.6 Handling Invalid Zero Values :-**

Zero values in Glucose, BloodPressure, SkinThickness, Insulin, and BMI are treated as missing measurements. These values are replaced with `NaN` and imputed using the median of each column to retain the records for further analysis.

***Replacing zeros with missing values***

```python
cols = ['Glucose', 'BloodPressure', 'SkinThickness',
        'Insulin', 'BMI']

df[cols] = df[cols].replace(0, np.nan)
```

***Filling missing values using the median***

```python
for col in cols:
    df[col] = df[col].fillna(df[col].median())
```

***Verifying the cleaning***

```python
print("Missing Values:\n", df.isnull().sum())
print("\nZero Values:\n", df[cols].eq(0).sum())
print("\nDataset Shape:", df.shape)
```

## **6. Exploratory Data Analysis (EDA)**

Exploratory Data Analysis is performed to understand the distribution of variables, identify patterns, and explore relationships between health-related measurements and the recorded diabetes outcome.

**6.1 Outcome Distribution :-**

This analysis examines the distribution of the Outcome variable to compare the number of records labeled as non-diabetic (0) and diabetic (1) in the dataset.

***Counting the outcomes***

```python
df['Outcome'].value_counts()
```

***Visualizing the outcomes***

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x='Outcome')

plt.title('Distribution of Diabetes Outcome')
plt.xlabel('Outcome (0 = No Diabetes, 1 = Diabetes)')
plt.ylabel('Number of Records')
plt.show()
```

***Calculating percentages***

```python
(df['Outcome'].value_counts(normalize=True) * 100).round(2)
```

**Insight:** Approximately 65.1% of records belong to Outcome 0 (no diabetes recorded), while 34.9% belong to Outcome 1 (diabetes recorded). This indicates that the dataset contains more non-diabetic than diabetic records.

**6.2 Glucose Distribution :-**

This analysis examines the distribution of glucose measurements to understand their spread, frequency, and central tendency in the dataset.

***Summary statistics***

```python
df['Glucose'].describe()
```

***Histogram***

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='Glucose', kde=True)

plt.title('Distribution of Glucose Levels')
plt.xlabel('Glucose Level')
plt.ylabel('Frequency')
plt.show()
```

***Comparing glucose by outcome***

```python
plt.figure(figsize=(7, 5))

sns.boxplot(data=df, x='Outcome', y='Glucose')

plt.title('Glucose Levels by Diabetes Outcome')
plt.xlabel('Outcome (0 = No Diabetes, 1 = Diabetes)')
plt.ylabel('Glucose Level')
plt.show()
```

**Insight:** The box plot shows that records with Outcome 1 have a higher median glucose level than records with Outcome 0. This suggests an association between higher glucose levels and the recorded diabetes outcome in this dataset. Some outliers are also present in the Outcome 0 group.

**6.6 Correlation Heatmap :-**

Correlation analysis measures the strength and direction of linear relationships between numerical variables. A heatmap is used to visualize these correlations, including their relationships with the Outcome variable.

```python
plt.figure(figsize=(10, 7))

correlation = df.corr(numeric_only=True)

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Diabetes Dataset')
plt.show()
```

***Correlation with Outcome***

```python
correlation['Outcome'].sort_values(ascending=False)
```

**Insight:** Glucose has the strongest positive correlation with Outcome (0.49), followed by BMI (0.31), Age (0.24), and Pregnancies (0.22). BloodPressure has a relatively weak positive correlation (0.17). These findings suggest that Glucose is the most strongly associated feature with the recorded diabetes outcome among the variables examined.
