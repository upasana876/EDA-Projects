**Insight:** The scatter plot shows a clear positive relationship between engine size and CO₂ emissions. Vehicles with smaller engines generally have lower emissions, while vehicles with larger engines tend to produce higher emissions. This suggests that engine size is an important factor associated with vehicle CO₂ emissions in this dataset. However, other vehicle characteristics may also influence emissions.

**6.19 Outlier Analysis :-**

Outlier analysis helps identify observations that differ substantially from the typical range of numerical values. The Interquartile Range (IQR) method is used to flag potential outliers in selected vehicle specifications, fuel consumption measurements, and CO₂ emissions.

***Box plots :***

```python
outlier_cols = [
    'ENGINESIZE',
    'CYLINDERS',
    'FUELCONSUMPTION_CITY',
    'FUELCONSUMPTION_HWY',
    'FUELCONSUMPTION_COMB',
    'CO2EMISSIONS'
]

fig, axes = plt.subplots(2, 3, figsize=(15, 8))

for col, ax in zip(outlier_cols, axes.flatten()):
    sns.boxplot(y=df[col], ax=ax)
    ax.set_title(col)

plt.tight_layout()
plt.show()
```

***Count of potential outliers using IQR :***

```python
outlier_summary = {}

for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outlier_summary[col] = (
        ((df[col] < lower) | (df[col] > upper)).sum()
    )

pd.Series(outlier_summary, name='Potential Outlier Count')
```

***Inspection of high CO₂ outliers :***

```python
Q1 = df['CO2EMISSIONS'].quantile(0.25)
Q3 = df['CO2EMISSIONS'].quantile(0.75)
IQR = Q3 - Q1

upper_limit = Q3 + 1.5 * IQR

df[df['CO2EMISSIONS'] > upper_limit][
    ['MAKE', 'MODEL', 'VEHICLECLASS',
     'ENGINESIZE', 'CYLINDERS', 'CO2EMISSIONS']
].sort_values('CO2EMISSIONS', ascending=False)
```
