## **5. Data Cleaning**

Data cleaning is an important step in EDA that helps improve data quality by identifying missing values, duplicate records, and incorrect data types.

### *5.1 Checking Missing Values*

The `isnull().sum()` function identifies the number of missing values in each column.

```python
df.isnull().sum()
```

### *5.2 Checking Duplicate Records*

The `duplicated().sum()` function counts the number of duplicate rows in the dataset.

```python
df.duplicated().sum()
```

### *5.3 Checking Data Types*

The data types of each column are checked to identify any columns that may require conversion.

```python
df.dtypes
```

### *5.4 Handling Missing Values*

Missing values are handled based on the nature and importance of each column. For categorical columns, missing values can be replaced with `"Unknown"` where appropriate.

#### *Step 1: Filling missing values*

```python
cols = ['director', 'cast', 'country', 'rating']

for col in cols:
    df[col] = df[col].fillna('Unknown')
```

#### *Step 2: Handle date_added*

```python
df['date_added'] = df['date_added'].fillna('Unknown')
```

#### *Step 3: Handle duration*

```python
df['duration'] = df['duration'].fillna('Unknown')
```

#### *Step 3: Check missing values again*

```python
df.isnull().sum()
```

### *5.5 Handling Duplicate Records*

Duplicate records are identified to avoid repeated entries that may affect the analysis. Exact duplicate rows are removed if present.

```python
df.duplicated().sum()
```

### *5.6 Final Data Quality Check*

The dataset is checked after cleaning to confirm its dimensions and verify that missing values and duplicate records have been handled.

```python
print("Dataset Shape:", df.shape)
print("\nMissing Values:\n", df.isnull().sum())
print("\nDuplicate Rows:", df.duplicated().sum())
```

## **6. Exploratory Data Analysis (EDA)**

Exploratory Data Analysis is performed to identify patterns, trends, and relationships in the Netflix dataset using statistical analysis and visualizations.

### *6.1 Movies vs. TV Shows*

This analysis compares the number of Movies and TV Shows available on Netflix using a count plot.

**Count the content types:**

```python
df['type'].value_counts()
```

**Creating a bar chart:**

```python
sns.countplot(data=df, x='type')
plt.title('Distribution of Movies and TV Shows on Netflix')
plt.xlabel('Content Type')
plt.ylabel('Count')
plt.show()
```

**Calculating percentages:**

```python
(df['type'].value_counts(normalize=True) * 100).round(2)
```

**Insight:** Movies make up the majority of titles in the Netflix dataset, while TV Shows account for a smaller proportion.

### *6.2 Content Ratings Distribution*

This analysis examines the distribution of content ratings to understand the type of audience Netflix content is intended for.

**Counting each rating:**

```python
df['rating'].value_counts().head(10)
```

**Visualizing the top 10 ratings**

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, y='rating',
              order=df['rating'].value_counts().head(10).index)
plt.title('Top 10 Content Ratings on Netflix')
plt.xlabel('Count')
plt.ylabel('Rating')
plt.show()
```

**Insight:** TV-MA is the most common content rating on Netflix, followed by TV-14 and TV-PG. This indicates that a large proportion of the content in the dataset is intended for mature audiences and older teenagers.

### *6.3 Top 10 Countries Producing Netflix Content*

This analysis identifies the top 10 countries with the highest number of titles available in the Netflix dataset.

**Counting countries**

```python
country_counts = df['country'].str.split(', ').explode()

country_counts = country_counts[
    country_counts != 'Unknown'
]

country_counts.value_counts().head(10)
```

**Creating a bar plot**

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x=country_counts.value_counts().head(10).values,
    y=country_counts.value_counts().head(10).index
)

plt.title('Top 10 Countries Producing Netflix Content')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.show()
```

**Insight:** The United States contributes the highest number of titles in the Netflix dataset, followed by India and the United Kingdom. This shows that the Netflix content library has a strong contribution from the US, with India also being a major content-producing country.

### *6.4 Content Added Over the Years*

This analysis examines the number of movies and TV shows added to Netflix each year to identify content addition trends.

**Preparing the year data**

```python
df['year_added'] = pd.to_datetime(
    df['date_added'], errors='coerce'
).dt.year
```

**Ploting the trend**

```python
year_counts = df['year_added'].value_counts().sort_index()

plt.figure(figsize=(12, 5))
plt.plot(year_counts.index, year_counts.values, marker='o')
plt.title('Netflix Content Added Over the Years')
plt.xlabel('Year')
plt.ylabel('Number of Titles Added')
plt.grid(True)
plt.show()
```

**Finding the peak year**

```python
year_counts.idxmax(), year_counts.max()
```

**Insight:** Netflix content additions increased significantly between 2016 and 2019, reaching a peak of 1,999 titles in 2019. After 2019, the number of titles added declined in the dataset.

### *6.5 Top 10 Genres on Netflix*

This analysis identifies the most common genres in the Netflix dataset by separating titles associated with multiple genres.

**Counting genres**

```python
genres = df['listed_in'].str.split(', ').explode()

genres.value_counts().head(10)
```

**Visualizing genres**

```python
plt.figure(figsize=(10, 5))

top_genres = genres.value_counts().head(10)

sns.barplot(x=top_genres.values, y=top_genres.index)

plt.title('Top 10 Genres on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.show()
```

**Insight:** International Movies is the most common genre in the Netflix dataset, followed by Dramas and Comedies. This indicates a diverse content library with a strong presence of international films and drama-based content.

### *6.6 Movies vs. TV Shows by Release Year*

This analysis compares the release-year trends of Movies and TV Shows to understand how their distributions have changed over time.

```python
plt.figure(figsize=(12, 5))

sns.histplot(
    data=df,
    x='release_year',
    hue='type',
    bins=30,
    multiple='stack'
)

plt.title('Movies vs. TV Shows by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.show()
```

**Insight:** The number of titles released increased significantly in recent years. Movies consistently outnumber TV Shows across most release years, with a noticeable concentration of titles released between 2015 and 2020.
