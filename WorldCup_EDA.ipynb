**6.2 Team Performance Analysis :-**

Analyzing match winners helps compare the number of recorded wins for participating teams. This analysis excludes ties, no-result matches, and abandoned matches from the team-win ranking.

***Bar chart :***

```python
special_outcomes = [
    'No result',
    'Match tied',
    'Match abandoned without a ball being bowled'
]

team_wins = df.loc[
    ~df['winner'].isin(special_outcomes),
    'winner'
].value_counts().head(10)

plt.figure(figsize=(10, 5))
team_wins.plot(kind='bar')

plt.title('Top 10 Teams by Recorded World Cup Wins')
plt.xlabel('Team')
plt.ylabel('Number of Wins')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Key Observations:**

- **Australia** has the highest number of recorded World Cup wins, with 69 wins.
- **New Zealand** ranks second with 54 wins, closely followed by **India** with 53 wins.
- England has 48 wins, followed by Pakistan with 45 wins.
- West Indies, Sri Lanka, and South Africa have 43, 40, and 38 wins, respectively.
- Bangladesh and Zimbabwe have fewer recorded wins among the top 10 teams, with 14 and 12 wins, respectively.

**Conclusion:** Australia leads the recorded match-win count in this dataset. These counts represent recorded wins in the available dataset, not the number of World Cup tournament titles won by each team.

**6.4 Team-wise Average Runs Analysis :-**

This analysis compares the average runs scored by each team across recorded World Cup matches. Scores are combined from both team positions to provide a broader view of team scoring performance. Matches with unavailable scores are excluded from the average calculation.

```python
team1_scores = df[['team_1', 'team_1_runs']].rename(
    columns={'team_1': 'team', 'team_1_runs': 'runs'}
)

team2_scores = df[['team_2', 'team_2_runs']].rename(
    columns={'team_2': 'team', 'team_2_runs': 'runs'}
)

all_team_scores = pd.concat(
    [team1_scores, team2_scores],
    ignore_index=True
)

team_avg_runs = (
    all_team_scores
    .dropna(subset=['runs'])
    .groupby('team')['runs']
    .mean()
    .sort_values(ascending=False)
)

team_avg_runs.head(10)
```

***Visualize the top 10 teams :***

```python
plt.figure(figsize=(10, 5))

team_avg_runs.head(10).sort_values().plot(kind='barh')

plt.title('Top 10 Teams by Average Runs Scored')
plt.xlabel('Average Runs')
plt.ylabel('Team')
plt.tight_layout()
plt.show()
```

**Key Observations:**

- Australia has the highest average score among the top 10 teams, with approximately **241.17 runs**.
- India ranks second with an average of **236.14 runs**.
- South Africa ranks third, averaging **232.70 runs**.
- England, Sri Lanka, and Pakistan have average scores of approximately 227.22, 224.90, and 224.51 runs, respectively.
- Zimbabwe has the lowest average score among these top 10 teams, at approximately **203.46 runs**.
- The average scores of the top 10 teams range from approximately 203 to 241 runs.

**Conclusion:** Australia has the highest average recorded score among the teams shown. However, average runs alone do not determine match success; winning frequency, opposition, match conditions, and tournament format also matter.
